1. Count Vowels

Write a function `count_vowels(s)` that takes a string and returns the number of vowels in it (`a`, `e`, `i`, `o`, `u` — both upper and lower case).

```python
>>> count_vowels("hello")
2
>>> count_vowels("HEllO World")
3

````python
### BEGIN SOLUTION

def count_vowels(s):
    """
    (str) -> int
    This function counts the number of vowels in a string,both the upper and
    lower cases.

    >>> count_vowels("hello")
    2
    >>> count_vowels("HELLO")
    2
    >>> count_vowels("xyz")
    0
    """
    count = 0

    for letter in s:
        if letter in "aeiouAEIOU":
            count += 1

    return count


### END SOLUTION
````

````python
assert count_vowels("hello") == 2
assert count_vowels("HELLO") == 2
assert count_vowels("xyz") == 0

````

2 BMI Calculator

Write a function bmi(weight_kg, height_m) that returns the Body Mass Index (BMI) using the formula:

$
BMI = \frac{\text{weight (kg)}}{(\text{height (m)})^2}
$

Include doctests for values like:
	•	weight_kg=70, height_m=1.75 → 22.86 (rounded to two decimals)

````python
from contextlib import nullcontext
### BEGIN SOLUTION
def bmi(weight, height):
  """
  This function calculates the Body Mass Index using the ratio of the weight in
  kilograms and the square of the height in meters of an individual.
  If the weight and height are less than or equal to zero, the code with return
  None.
  This code will also round the the bmi to the two decimal places

  >>> bmi(70,1.75)
  22.86
  >>> bmi(50,1.6)
  19.53
  """
  if weight > 0 and height > 0:
    return round(weight/ height/height, 2)
  else:
    return None

### END SOLUTION
````

3 Filter Short Words

Create a function filter_short(words, max_len) that returns a list of only those words whose length is less than or equal to max_len.
```python
>>> filter_short(['hi', 'hello', 'no', 'yes'], 2)
['hi', 'no']

````python

def filter_short(words, max_len):
    """
    This function returns a list of words whose length
    is less than or equal to max_len.

    >>> filter_short(['hi', 'hello', 'no', 'yes'], 2)
    ['hi', 'no']

    >>> filter_short(['apple', 'banana', 'pear'], 4)
    ['pear']
    """
    short_words = []

    for word in words:
        if len(word) <= max_len:
            short_words += [word]

    return short_words
````

````python
assert filter_short(['hi', 'hello', 'no', 'yes'], 2) == ['hi', 'no']
assert filter_short(['apple', 'banana', 'pear'], 4) == ['pear']

````
